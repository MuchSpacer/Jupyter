```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams['axes.xmargin'] = 0
```

```python
N = 500
A = 5
B = 7
L = 1
```

```python
Psi_x = lambda x, a: (2/L) * np.sin((a*np.pi/L)*(x-L/2))
Psi_y = lambda y, b: np.sin((b*np.pi/L)*(y-L/2))
```

```python
X = np.linspace(-L/2, L/2, N)
# Y array is equal to X
X_d = np.arange(0,N)
```

```python
Z1_x = np.full((N,N), Psi_x(X,A))
Z1_y = np.full((N,N), Psi_y(X,B))
Z2_x = np.full((N,N), Psi_x(X,B))
Z2_y = np.full((N,N), Psi_y(X,A))
Z = ((Z1_x * Z1_y.T) - (Z2_x * Z2_y.T)) / np.sqrt(2)
```

```python
Ticks = [-L/2, -L/4, 0, L/4, L/2]
Labels = [r"$-L/2$", r"$-L/4$", "$0$", r"$L/4$", r"$L/2$"]

plt.contourf(X, X, Z, cmap = "coolwarm")
plt.gca().set_aspect('equal')
plt.xticks(Ticks, Labels)
plt.yticks(Ticks, Labels)
plt.title(r"$\Psi_{-} = \frac{1}{\sqrt{2}}(\Psi_{5,7}-\Psi_{7,5})$ in Infinite Square Well")
plt.colorbar()
plt.contour(X, X, Z, colors = "black", linewidths = 1)
plt.savefig("HW02_ISW.png")
```
