```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['axes.xmargin'] = 0
```

```python
x = np.linspace(0,0.5,150)
```

```python
M1=np.exp(0.5)/(np.exp(0.5)-np.exp(-0.5))
M2=np.exp(-0.5)/(np.exp(0.5)-np.exp(-0.5))
```

```python
y = M1 * np.exp(-x) - M2 * np.exp(x)
y2 = 1 - 2 * x
```

```python
plt.plot(x,y, label = "Actual Carrier Concentration")
plt.plot(x,y2, "--", label = "Linear Approximation")
plt.grid()
plt.ylim(0,1.05)
plt.legend()
plt.ylabel(r"$\delta_p / \Delta_{pE}$")
plt.xlabel(r"$x_n / L_p$")
plt.savefig("HW10Problem2.png")
```
